```python
import pandas as pd
```

```python
# Load the structured HDFS dataset

df = pd.read_csv("../datasets/HDFS_2k_structured.csv")

print("Dataset Shape:", df.shape)

df.head()
```

```python
# Feature 1: Message Length

df["Message_Length"] = df["Message"].str.len()

df[["Message", "Message_Length"]].head()
```

## Observation: Message Length Feature

A new feature called **Message_Length** was created by calculating the number of characters in each log message.

This converts raw textual information into a numerical feature that can be used by machine learning algorithms. Longer or unusually short log messages may indicate abnormal system behaviour, making this feature potentially useful for anomaly detection.


```python
# Feature 2: Block ID Presence

df["Has_Block_ID"] = df["Message"].str.contains("blk_", regex=False).astype(int)

df[["Message", "Has_Block_ID"]].head()
```

## Observation: Block ID Presence Feature

A binary feature named **Has_Block_ID** was created to indicate whether a log message contains a Hadoop block identifier (`blk_`).

Representing this information numerically allows machine learning models to use the presence or absence of block identifiers as an additional feature during anomaly detection.

```python
# Feature 3: PacketResponder Presence

df["Has_PacketResponder"] = df["Message"].str.contains("PacketResponder", case=False).astype(int)

df[["Message", "Has_PacketResponder"]].head()
```

## Observation: PacketResponder Feature

A binary feature named **Has_PacketResponder** was created to identify whether a log message contains the keyword **PacketResponder**.

PacketResponder is a Hadoop DataNode component responsible for responding to data packet operations. Encoding its presence as a binary feature may help machine learning models distinguish between different categories of system events.

```python
# Feature 4: Error Keyword Presence

df["Has_Error"] = df["Message"].str.contains("error", case=False, na=False).astype(int)

df[["Message", "Has_Error"]].head()
```

## Observation: Error Keyword Feature

A binary feature named **Has_Error** was created to identify whether a log message contains the keyword **error**.

Although the first few HDFS log messages do not contain this keyword, such messages may appear later in the dataset. Including this feature allows anomaly detection models to recognize explicit error-related events.

```python
# Feature 5: Warning Log

df["Is_WARN"] = (df["Level"] == "WARN").astype(int)

df[["Level", "Is_WARN"]].head()
```

## Observation: Warning Log Feature

A binary feature named **Is_WARN** was created based on the log level.

The feature assigns a value of **1** to warning messages and **0** to informational messages. Since warning events may indicate abnormal system behaviour, this feature is expected to contribute to anomaly detection performance.

# Feature Engineering Summary

Five new features were successfully engineered from the structured HDFS log dataset.

The engineered features include:

- Message_Length
- Has_Block_ID
- Has_PacketResponder
- Has_Error
- Is_WARN

These features transform raw log messages into numerical and binary representations that can be used as inputs for anomaly detection algorithms in subsequent experiments.
