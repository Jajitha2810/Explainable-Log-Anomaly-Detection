```python
import pandas as pd

data = []

for line in logs:
    parts = line.split()

    data.append({
        "Date": parts[0],
        "Time": parts[1],
        "Thread_ID": parts[2],
        "Level": parts[3],
        "Component": parts[4],
        "Message": " ".join(parts[5:])
    })

df = pd.DataFrame(data)

df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
level_counts = df["Level"].value_counts()

plt.figure(figsize=(6,4))
level_counts.plot(kind="bar")

plt.title("Distribution of Log Levels")
plt.xlabel("Log Level")
plt.ylabel("Count")

plt.show()
```

```python
level_counts = df["Level"].value_counts()

plt.figure(figsize=(6,4))
level_counts.plot(kind="bar")

plt.title("Distribution of Log Levels")
plt.xlabel("Log Level")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig("../figures/log_level_distribution.png", dpi=300)

plt.show()
```

## Observation: Log Level Distribution

The HDFS dataset contains two log levels: INFO and WARN.

- INFO logs: 1920
- WARN logs: 80

The graph shows that INFO logs dominate the dataset, representing the vast majority of log entries. WARN logs occur much less frequently, indicating that warning events are relatively rare in this dataset.

This observation suggests that most system activities recorded in the HDFS logs correspond to normal operational events.

```python
component_counts = df["Component"].value_counts()

plt.figure(figsize=(8,5))
component_counts.plot(kind="bar")

plt.title("Distribution of Hadoop Components")
plt.xlabel("Component")
plt.ylabel("Count")

plt.xticks(rotation=30, ha="right")

plt.tight_layout()

plt.savefig("../figures/component_distribution.png", dpi=300)

plt.show()
```

## Observation: Hadoop Component Distribution

The HDFS logs contain messages from multiple Hadoop components.

The most frequently occurring component is **dfs.FSNamesystem**, followed by **dfs.DataNode$PacketResponder** and **dfs.DataNode$DataXceiver**. Components such as **dfs.FSDataset** appear less frequently, while **dfs.DataBlockScanner** and **dfs.DataNode** occur only rarely.

This distribution indicates that most log messages are generated by core HDFS operations involving block management and data transfer.
