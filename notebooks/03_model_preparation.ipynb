```python
import pandas as pd
```

```python
# Load structured dataset

df = pd.read_csv("../datasets/HDFS_2k_structured.csv")

df.head()
```

```python
# Feature Engineering

df["Message_Length"] = df["Message"].str.len()

df["Has_Block_ID"] = df["Message"].str.contains(
    "blk_", regex=False
).astype(int)

df["Has_PacketResponder"] = df["Message"].str.contains(
    "PacketResponder",
    case=False
).astype(int)

df["Has_Error"] = df["Message"].str.contains(
    "error",
    case=False,
    na=False
).astype(int)

df["Is_WARN"] = (df["Level"] == "WARN").astype(int)

df.head()
```

```python
# Selecting features for machine learning

features = [
    "Message_Length",
    "Has_Block_ID",
    "Has_PacketResponder",
    "Has_Error",
    "Is_WARN"
]

X = df[features]

X.head()
```

## Observation: Machine Learning Feature Matrix

A feature matrix named **X** was created by selecting the engineered numerical and binary features.

The feature matrix excludes raw textual information and retains only machine-learning-ready variables.

This matrix will serve as the input for anomaly detection algorithms in the subsequent experiments.
