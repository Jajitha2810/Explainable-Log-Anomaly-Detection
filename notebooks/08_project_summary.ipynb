# Explainable Log Anomaly Detection Using Machine Learning

## Project Summary

This notebook provides a complete overview of the Explainable Log Anomaly Detection project, including the research objective, dataset, feature engineering process, machine learning model, evaluation results, explainability analysis, and final conclusions.

# Research Objective

The objective of this research is to develop an explainable machine learning framework capable of detecting anomalous events within Hadoop Distributed File System (HDFS) log data.

Unlike traditional anomaly detection systems, this project integrates Explainable Artificial Intelligence (XAI) techniques using SHAP (SHapley Additive exPlanations) to improve the transparency and interpretability of anomaly predictions.

# Dataset

Dataset Used:
- HDFS Log Dataset

Dataset Size:
- 2,000 structured log entries

Each log contains information such as:

- Date
- Time
- Thread ID
- Log Level
- Component
- Log Message

# Feature Engineering

Five features were engineered from the original log messages:

1. Message_Length
2. Has_Block_ID
3. Has_PacketResponder
4. Has_Error
5. Is_WARN

These features were selected because they capture structural and semantic characteristics commonly associated with abnormal HDFS behaviour.

# Machine Learning Model

Model Used:
- Isolation Forest

Parameters:

- contamination = 0.05
- random_state = 42

Isolation Forest was selected because it is an unsupervised anomaly detection algorithm that performs well on datasets where labelled anomalies are unavailable.

# Experimental Results

Dataset Size:
- 2,000 logs

Detected Anomalies:
- 88 logs

Normal Logs:
- 1,912 logs

The model successfully identified approximately 4–5% of the dataset as anomalous.

# Explainability Results

SHAP analysis revealed that the most influential features were:

1. Message_Length
2. Has_PacketResponder
3. Is_WARN
4. Has_Error
5. Has_Block_ID

Among these, Message_Length contributed the most towards anomaly predictions, indicating that longer operational log messages were more likely to represent unusual system behaviour.

# Key Findings

- Isolation Forest successfully detected anomalous HDFS log events.
- The DataNode DataXceiver component produced the highest number of anomalous logs.
- Anomalous log messages were significantly longer than normal log messages.
- SHAP explainability improved model transparency by identifying the features contributing most strongly to anomaly predictions.

This project successfully demonstrates the application of Explainable Artificial Intelligence to HDFS log anomaly detection.
